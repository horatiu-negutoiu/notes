```python
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

# Download and prepare data
print(os.getcwd())
data_root = "data/"
train_data = pd.read_csv(data_root + "train.csv")
x_train = train_data[["x"]]
y_train = train_data["y"]
```

```python
# Visualize the training data
train_data.plot(kind='scatter', grid=True, x='x', y='y')
plt.axis([0, 100, -4, 110])
plt.show()
```

```python
# Select and train linear model
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
# Predict a new value
x_new = [[80]]
y_new = model.predict(x_new)
print(f'Prediction for x={x_new[0][0]}: y={y_new[0]}')
```

```python
# Predict test values and evaluate the model performance

from sklearn.metrics import mean_squared_error, r2_score

test_data = pd.read_csv(data_root + "test.csv")
x_test = test_data[["x"]]
y_test = test_data["y"]

y_pred = model.predict(x_test)

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Mean Squared Error:", mse)
print("R² Score:", r2)

print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
```
